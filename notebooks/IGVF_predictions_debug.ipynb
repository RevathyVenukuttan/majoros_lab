```python
%%writefile /hpc/home/rv103/igvf/revathy/IGVF_predictions/scripts/validate_predictions.py
# validate_predictions_stream.py
import argparse

EXPECTED_ALLELES = ("A", "C", "G", "T")
EXPECTED_SET = set(EXPECTED_ALLELES)

def parse_region_id(region_id: str):
    # "chr1:104822-105122" -> ("chr1", 104822, 105122)
    chrom, rest = region_id.split(":", 1)
    start_s, end_s = rest.split("-", 1)
    return chrom, int(start_s), int(end_s)

def main():
    ap = argparse.ArgumentParser()
    ap.add_argument("predictions", help="predictions.txt (TSV)")
    ap.add_argument("--missing-out", default="missing_entries.tsv",
                    help="Where to write missing (ID, pos, missing_alleles)")
    ap.add_argument("--max-missing-lines", type=int, default=50000,
                    help="Cap lines written to missing-out to avoid huge output")
    ap.add_argument("--check-300bp", action="store_true",
                    help="Also verify each ID covers exactly 300 positions (start..end-1)")
    args = ap.parse_args()

    total_rows = 0
    total_positions = 0
    total_missing_alleles = 0
    bad_rows = 0

    current_id = None
    current_pos = None
    seen_alleles = set()

    # For per-ID checks
    id_start = None
    id_end = None
    seen_positions_in_id = set() if args.check_300bp else None

    written_missing_lines = 0

    def flush_position(out_f):
        """Called when we finish a (current_id, current_pos) group."""
        nonlocal total_positions, total_missing_alleles, written_missing_lines
        if current_id is None or current_pos is None:
            return
        total_positions += 1
        missing = EXPECTED_SET - seen_alleles
        if missing:
            total_missing_alleles += len(missing)
            if written_missing_lines < args.max_missing_lines:
                out_f.write(f"{current_id}\t{current_pos}\t{','.join(sorted(missing))}\n")
                written_missing_lines += 1

    def flush_id_checks():
        """Optional: verify that an ID has exactly 300 positions."""
        nonlocal bad_rows
        if not args.check_300bp or current_id is None:
            return
        # Expected positions are start..end-1 (because 105122-104822 = 300)
        exp_positions = (id_end - id_start)
        if exp_positions != 300:
            bad_rows += 1  # unexpected interval length
        if len(seen_positions_in_id) != 300:
            # This indicates missing/duplicate positions within this ID.
            bad_rows += 1

    with open(args.predictions, "rt", encoding="utf-8", newline="") as f, \
         open(args.missing_out, "wt", encoding="utf-8") as out_f:

        header = f.readline()
        out_f.write("ID\tpos\tmissing_alleles\n")

        for line_num, line in enumerate(f, start=2):
            line = line.rstrip("\n")
            if not line:
                continue

            parts = line.split("\t")
            if len(parts) < 6:
                bad_rows += 1
                continue

            rid = parts[0]
            pos_field = parts[2]     # like "pos=104822"
            allele = parts[4]

            # Parse pos
            if not pos_field.startswith("pos="):
                bad_rows += 1
                continue
            try:
                pos = int(pos_field[4:])
            except ValueError:
                bad_rows += 1
                continue

            # If new ID, flush last position + ID checks
            if rid != current_id:
                flush_position(out_f)
                flush_id_checks()

                current_id = rid
                current_pos = None
                seen_alleles.clear()

                if args.check_300bp:
                    _, id_start, id_end = parse_region_id(rid)
                    seen_positions_in_id.clear()

            # If new position, flush previous position
            if pos != current_pos:
                flush_position(out_f)
                current_pos = pos
                seen_alleles.clear()

            # Track
            total_rows += 1
            seen_alleles.add(allele)
            if args.check_300bp:
                seen_positions_in_id.add(pos)

            # (Optional) lightweight progress print
            if total_rows % 50_000_000 == 0:
                print(f"Processed {total_rows:,} rows...")

        # Flush final groups
        flush_position(out_f)
        flush_id_checks()

    print("Done.")
    print(f"Total rows processed: {total_rows:,}")
    print(f"Total positions seen: {total_positions:,}")
    print(f"Total missing allele calls (A/C/G/T) across positions: {total_missing_alleles:,}")
    print(f"Bad/malformed rows: {bad_rows:,}")
    print(f"Missing report written to: {args.missing_out}")
    if written_missing_lines >= args.max_missing_lines:
        print(f"NOTE: missing-out capped at {args.max_missing_lines} lines.")

if __name__ == "__main__":
    main()
```
