```python
import pandas as pd
import vcfpy
import sys
import gzip


with gzip.open("/hpc/home/rv103/igvf/IGVF-DACC/IGVFFI2231RETG.vcf.gz", "r") as infile, open('/hpc/home/rv103/igvf/revathy/IGVF_predictions/data/Y2AVE_variants.txt',"wt") as outfile:
    for line in infile:
        line = line.decode("utf-8")
        if line.startswith("#") == False:
            record = line.strip().split("\t")
            chrom = record[0]
            pos = str(int(record[1])-1) # convert 1-based index to 0-based index to fit 2bittofa
            ref = record[3].upper()
            alt = record[4].upper()
            out = chrom+':'+pos+':'+'ref='+ref+':'+ref+','+alt+'\n'
            outfile.write(out)

    outfile.close()
```

```python
%%writefile /hpc/home/rv103/igvf/revathy/notebooks/runtime.py
import gzip
from collections import defaultdict

vcf_path = "/hpc/home/rv103/igvf/IGVF-DACC/IGVFFI2231RETG.vcf.gz"

# Store: key = (chrom, pos, ref), value = set of ALT alleles observed
site_alts = defaultdict(set)

with gzip.open(vcf_path, "rt") as infile:
    for line in infile:
        if line.startswith("#"):
            continue

        record = line.strip().split("\t")

        chrom = record[0]
        pos = record[1]             # keep as string
        ref = record[3].upper()
        alts = record[4].upper().split(",")  # should be single alt per row in your file

        key = (chrom, pos, ref)

        # Add all ALT alleles (even though each row should contain only 1)
        for a in alts:
            if a != ".":
                site_alts[key].add(a)

# ---- Counting ----

num_unique_variants = len(site_alts)
ref_count = num_unique_variants               # 1 REF per unique site
alt_count = sum(len(a) for a in site_alts.values())

print("Unique variant sites:", num_unique_variants)
print("Total REF alleles:", ref_count)
print("Total ALT alleles:", alt_count)

```
